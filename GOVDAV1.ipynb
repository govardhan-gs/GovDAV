```python
import pandas as pd

# 1. Create a sample DataFrame
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": 45,
    "City": ["New York", "Paris", "London", "Tokyo", "Berlin"],
}
# Set specific strings as row labels (index)
df = pd.DataFrame(data, index=["row1", "row2", "row3", "row4", "row5"])

print("--- Original DataFrame ---")
print(df)
print("\n")


# 2. Slicing by Integer Position using .iloc
# Syntax: df.iloc[row_start:row_end, col_start:col_end]
# Note: The stop index is excluded (just like standard Python lists)
print("--- Slicing with .iloc (Rows 1 to 3, Columns 0 to 2) ---")
iloc_sliced = df.iloc[1:4, 0:2]
print(iloc_sliced)
print("\n")


# 3. Slicing by Label/Name using .loc
# Syntax: df.loc[row_start:row_end, col_start:col_end]
# Note: The stop index is included
print("--- Slicing with .loc (Rows 'row2' to 'row4', Columns 'Name' to 'Age') ---")
loc_sliced = df.loc["row2":"row4", "Name":"Age"]
print(loc_sliced)
print("\n")


# 4. Standard Bracket Slicing (Rows only)
# Shortcut to slice rows by integer positions without typing .iloc
print("--- Slicing rows directly with brackets (Rows 0 to 2) ---")
bracket_sliced = df[0:3]
print(bracket_sliced)
print("\n")


# 5. Slicing with Steps (Skipping rows)
# Uses the [start:stop:step] pattern to select every second row
print("--- Slicing with a step of 2 (Every other row) ---")
step_sliced = df.iloc[0:5:2, :]
print(step_sliced)
```
